## Setup

```python
bks_df = df.copy()
```

### General renames

```python
bks_df = bks_df.rename(
    columns={
        "Which describes you best? (cvc5b81)": "ds_preference",
        "Which of the following is the most severe for you? (trspees)": "diagnosis",
        "Your sexual interests feel (44qhm16)": "sexual_interest_variety",
    }
)
```

### OCEAN cleanup
Cast OCEAN columns to int and strip the `variable` suffix so they can be attached as numeric keys.

```python
ocean_cols = [
    "opennessvariable",
    "consciensiousnessvariable",
    "extroversionvariable",
    "neuroticismvariable",
    "agreeablenessvariable",
    "powerlessnessvariable",
]
bks_df[ocean_cols] = bks_df[ocean_cols].astype(int)
bks_df = bks_df.rename(columns={c: c.replace("variable", "") for c in ocean_cols})
ocean_cols = [c.removesuffix("variable") for c in ocean_cols]
```

### Attribute keys to attach to SFW/NSFW
Personal attributes merged into each sub-df so preferences can be reviewed against demographics, personality, D/s and CGL signal without joining back to `bks_df`.

```python
attach_cols = [
    "respondent_id",
    # demographics
    "age",
    "biomale",
    "straightness",
    "politics",
    # behavioral
    "sexcount",
    "pornhabit",
    # CGL signal
    "cgl",
    # D/s preference
    "ds_preference",
    '"I am aroused by being dominant in sexual interactions" (6w3xquw)',
    '"I am aroused by being submissive in sexual interactions" (xem7hbu)',
    # personality (Big Five + powerlessness)
    *ocean_cols,
    # clinical / self-assessment
    "diagnosis",
    "sexual_interest_variety",
]
bks_df[attach_cols].head()
```

### Multi-select expansion helpers
Avoid column-name explosion in `bks_df` by routing expanded multi-selects to side-dataframes.

```python
# Patterns to drop from any expanded multi-select dummy column names
EXCLUDE_PATTERNS = (
    "incest", "parent", "sibling", "grandparent",
    "aunt", "uncle", "niece", "nephew", "cousin",
    "13-17", "0-14", "ageplay",
)

def explode_multi(col, prefix, source=None):
    """Turn a comma-separated multi-select column into prefixed binary dummies."""
    source = bks_df if source is None else source
    return source[col].str.lower().str.get_dummies(sep=",").add_prefix(prefix)

def drop_excluded(dummies):
    """Filter out dummy columns whose names reference excluded themes (incest, underage)."""
    bad = [c for c in dummies.columns
           if any(p in c.lower() for p in EXCLUDE_PATTERNS)]
    return dummies.drop(columns=bad)

def attach_keys(sub_df):
    """Attach personal-attribute keys from bks_df so the sub-df is self-sufficient for analysis."""
    return sub_df.merge(bks_df[attach_cols], on="respondent_id", how="left")
```

## SFW & NSFW Export

### `sfw_preferences_df` — schema-safe column names
Multi-selects whose dummy column names are tame enough to share on standard social platforms.

```python
sfw_preferences_df = (
    bks_df[["respondent_id"]]
    .copy()
    .join(drop_excluded(explode_multi(
        "Which of the following body weights do you find most erotic? (1jq57nf)", "bodyweight_")))
    .join(drop_excluded(explode_multi(
        "Which of the following body types do you find significantly erotic?", "bodytype_")))
    .join(drop_excluded(explode_multi(
        "Check all that apply: Scenarios you find erotic tend to involve *you* feeling", "feel_self_")))
    .join(drop_excluded(explode_multi(
        "Check all that apply: Scenarios you find erotic tend to involve *the other person/people/creatures* feeling",
        "feel_other_")))
)
sfw_preferences_df = attach_keys(sfw_preferences_df)
sfw_preferences_df.shape
```

### `nsfw_preferences_df` — schema-explicit column names
Multi-selects whose dummy column names contain explicit kink/anatomy labels — only share on NSFW-permitting platforms.

```python
nsfw_preferences_df = (
    bks_df[["respondent_id"]]
    .copy()
    .join(drop_excluded(explode_multi(
        "Which of the following sexual positions do you find significantly erotic?", "pos_")))
    .join(drop_excluded(explode_multi(
        "Which of the following sex acts do you find significantly erotic?", "act_")))
    .join(drop_excluded(explode_multi(
        "Select all body parts you find significantly erotic", "bodypart_")))
    .join(drop_excluded(explode_multi(
        "Which of the following body modifications do you find significantly erotic?", "bodymod_")))
    .join(drop_excluded(explode_multi(
        "Common things: Check all the following categories that contain a thing that arouses you.", "common_")))
    .join(drop_excluded(explode_multi(
        "Uncommon things: Check all the following categories that contain a thing that arouses you.", "uncommon_")))
)
nsfw_preferences_df = attach_keys(nsfw_preferences_df)
nsfw_preferences_df.shape
```

### CGL Data Extraction

#### CGL review

```python
# Arousal-scale distribution in the cgl interest field, including NaN.
cgl_df.cgl.value_counts(dropna=False)
```

#### Roles extraction

```python
# Erotic role field extraction.
roles_col = "Which of the following roles are erotic?"
roles_df = cgl_df[roles_col].str.get_dummies(sep=",")

# Optional: namespace columns to avoid collisions, then join back
roles_df = roles_df.add_prefix("role_")
cgl_df = cgl_df.join(roles_df)
cgL_role_list = cgl_df.filter(like="role_").columns.tolist()

# Review
cgl_df.filter(like="role_").apply(pd.Series.value_counts).astype(int)
```

#### Soft erotic extraction

```python
# Soft erotic field extraction.
soft_erotic_col = "Which of the following do you find erotic? (vqogs86)"
soft_erotic_df = cgl_df[soft_erotic_col].str.get_dummies(sep=",")
soft_erotic_df = (
    cgl_df[soft_erotic_col]
    .str.lower()
    .str.get_dummies(sep=",")
    .add_prefix("soft_erotic_")
)
soft_erotic_df = soft_erotic_df.add_prefix("soft_erotic_")
cgl_df = cgl_df.join(soft_erotic_df)
soft_erotic_list = cgl_df.filter(like="soft_erotic_").columns.tolist()

cgl_df.columns = cgl_df.columns.str.replace(
    "soft_erotic_soft_erotic_", "soft_erotic_", regex=False
)
soft_erotic_list = cgl_df.filter(like="soft_erotic_").columns.tolist()

# Review
cgl_df.filter(like="soft_erotic_").apply(pd.Series.value_counts).fillna(0).astype(int)
```

#### Nonstandard age tag

```python
# Nonstandard age arousal field extraction based on a checklist of interests.
# This is a bit hacky, but it allows us to extract a signal about nonstandard age arousal without relying on the direct "cgl" field
# May be underreported or misunderstood by respondents.
uncommon_col = "Uncommon things: Check all the following categories that contain a thing that arouses you."
target = "Age: nonstandard (age gaps, ageplay, unusual ages, etc"

cgl_df["nonstandard_age_tag"] = np.where(
    cgl_df[uncommon_col].isna(),
    np.nan,
    cgl_df[uncommon_col].str.contains(target, regex=False, na=False).astype(float),
)

cgl_df["nonstandard_age_tag"].value_counts(dropna=False)
```

#### Build final_col

```python
# CGL-curated column list for the cgl_BKS_data.csv export.
cgL_role_list = cgl_df.filter(like="role_").columns.tolist()

final_col = [
    "respondent_id",
    "age",
    "cgl",
    *ocean_cols,
    "diagnosis",
    "sexual_interest_variety",
    "ds_preference",
    "youfeelmost",
    "otherfeel1most",
    "older",
    "agegap",
    "progression",
    "regression",
    "roles",
    "nonstandard_age_tag",
    *cgL_role_list,
    *soft_erotic_list,
]
```
